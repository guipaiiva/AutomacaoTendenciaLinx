```python
from selenium import webdriver
import os
from selenium.webdriver.common.by import By

driver = webdriver.Chrome()
driver.get("https://www.google.com")
driver.quit()

# pega o caminho da pasta onde está o .ipynb
caminho = os.getcwd()

# monta o caminho correto do arquivo
arquivo = os.path.join(caminho, "Arquivos/Pagina Hashtag.html")

# adiciona o prefixo file://
arquivo = f"file://{arquivo}"

navegador = webdriver.Chrome()
navegador.get(arquivo)


```

```python
# navegador.find_element traz 1 item
# navegador.find_elements Traz uma lista

navegador.find_element(By.ID, "fullname").send_keys("Guilherme Paiva")
navegador.find_element(By.ID, "email").send_keys("paivateste0@gmail.com")
navegador.find_element(By.ID, "_form_176_submit").click()

 
# Sempre procuro primeiro ID, depois a classe e por ultimo o xpath. 
```

```python
# Outras formas de pegar elementos Tag, Link_Text e name

#Pegando o texto que tem na tag h2
navegador.find_element(By.TAG_NAME, "h2").text

#Procura no texto escrito a palavra chave e traz o texto. Letra A a faz diferenca
numero_whatsapp = navegador.find_element(By.PARTIAL_LINK_TEXT, "WhatsApp").text
print(numero_whatsapp)

#procurar parametro name igual ID 
navegador.find_element(By.NAME, "email").send_keys("paivatestes0@gmail.com")


```

```python
#Usar o find elements -> trazer varios dados
lista_elementos = navegador.find_elements(By.CLASS_NAME, "nav-link" )
for elemento in lista_elementos:
   if "blog" in elemento.text.lower():
      elemento.click()
      break
```

```python
# Pegar o atributo de um elemento exemplo o href

texto = navegador.find_element(By.XPATH, "/html/body/footer/div/div[1]/div[2]/div/a[2]").get_attribute("href")
print(texto)

link  = navegador.find_element(By.XPATH, "/html/body/section[2]/div/div[8]/figure/a").get_attribute("href")
print(link)
```

```python
# Vou procurar no site inteiro pela tag name figure
lista_elementos2 = navegador.find_elements(By.TAG_NAME, "figure")
# usei o len para ver quantas tags figure tinha total 6
print(len(lista_elementos2))

for elemento in lista_elementos2:
        #O try é para ele tentar, caso nao encontre ele nao da erro ele so continua
        try:
            link = elemento.find_element(By.TAG_NAME, "a").get_attribute("href")
            print(link)
        except:
              continue    
   

```
