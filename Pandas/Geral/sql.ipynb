```python
#biblioteca pyodbc para conectar com banco de dados
# usa o comanda print(pyodbc.drivers()) para ver qual driver de banco de dados tem instalado no pc
import pyodbc

dados_conexao = "DRIVER=SQLite3;DATABASE=salarios.sqlite"

conexao = pyodbc.connect(dados_conexao)
cursor = conexao.cursor()

#Executar comando SQL select para puxar dados do banco de dados
cursor.execute("SELECT * FROM Salaries")
# Aqui pega as informacoes que ele pegou do banco de dados e jogou na variavel valores
valores = cursor.fetchall()


print(valores[:10])




cursor.close()
conexao.close()

```
